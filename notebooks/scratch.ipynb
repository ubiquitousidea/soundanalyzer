```python
import numpy as np
import pandas as pd
import plotly.graph_objs as go
import plotly.express as px
import json
```

```python
with open('pr_data.json', 'r') as f:
    d = json.load(f)
```

```python
t = [item['created_at'] for item in d]
```

```python
df = pd.DataFrame()
```

```python
df['t'] = t
```

```python
df['t'] = pd.to_datetime(df['t'])
```

```python
df['n'] = 1
```

```python
df = df.set_index('t')
```

```python
df = df.sort_index(ascending=True)
```

```python
df2 = df.resample('1H').count()
df2 = df2[df2['n'] > 0]
```

```python
df2 = df2.assign(n2 = df2['n'].cumsum())
```

```python
df2
```

```python
px.line(df2.reset_index(), x='t', y='n')
```

```python
px.line(df2.reset_index(), x='t', y='n2')
```

```python
df2 = df2.drop(['n2'], axis=1)
```

```python
df3 = df2[df2.n < 50]
```

```python
px.line(df3.reset_index(), x='t', y='n')
```

```python
df3 = df3.assign(dow = df3.index.dayofweek)
```

```python
px.histogram(df3, x='dow')
```

```python
avg1 = df3.groupby('dow').agg({'n': 'mean'})
```

```python
px.scatter(avg1.reset_index(), x='dow', y='n')
```
